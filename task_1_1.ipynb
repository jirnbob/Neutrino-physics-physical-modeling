```python
import numpy as np
import k3d
```

```python
#1)Детектор и положения ФЭУ
class Detector:
    def __init__(self, R=2.0, N=40, r=0.1, a=0.5):
        self.R = R
        self.N = N
        self.r = r
        self.a = a
        self.fmt_positions = self.generate_fmts()

    def generate_fmts(self):
        phi = np.random.uniform(0, 2 * np.pi, self.N)
        U = np.random.uniform(-1, 1, self.N)
        sin_theta = np.sqrt(1 - U**2)

        x = self.R * sin_theta * np.cos(phi)
        y = self.R * sin_theta * np.sin(phi)
        z = self.R * U

        return np.column_stack([x, y, z])

    #2)Проверка попадания на ФЭУ
    def find_fmt(self, point):
        d = np.sqrt(np.sum((self.fmt_positions - point)**2, axis=1))
        i = np.argmin(d)

        if d[i] <= self.r:
            return int(i)
        return -1
```

```python
#3)Параметры детектора
R = 2.0
N = 40
r = 0.1
a = 0.5

det = Detector(R, N, r, a)
```

```python
#4)Визуализация ФЭУ
plot = k3d.plot(name='Детектор')
fmt = k3d.points(det.fmt_positions.astype(np.float32), point_size=r, color=0x000000)
plot += fmt
plot.display()
```
