```python
import numpy as np
```

```python
# 1) Источник и генерация фотонов
class Source:
    def __init__(self, Y=10000, pos=(0.0, 0.0, 0.0)):
        self.Y = Y
        self.pos = np.array(pos, dtype=float)

    def generate_photons(self, E, batch_size=100):
        N = np.random.poisson(self.Y * E)

        while N > 0:
            n = min(batch_size, N)

            phi = np.random.uniform(0, 2 * np.pi, n)
            U = np.random.uniform(-1, 1, n)
            sin_theta = np.sqrt(1 - U**2)

            x = sin_theta * np.cos(phi)
            y = sin_theta * np.sin(phi)
            z = U

            N -= n
            yield np.column_stack([x, y, z])
```

```python
# 2) Параметры источника
Y = 10000
E = 1.0

source = Source(Y, pos=(0.0, 0.0, 0.0))
```

```python
# 3) Проверка генератора
N_photons = 0
N_batches = 0
n_max = 0

for directions in source.generate_photons(E):
    n = len(directions)
    N_photons += n
    N_batches += 1
    n_max = max(n_max, n)

print(f"Всего фотонов: {N_photons}")
print(f"Ожидаемое число фотонов: {Y * E:.0f}")
print(f"Количество пачек: {N_batches}")
print(f"Максимальный размер пачки: {n_max}")
```

```python
# 4) Проверка распределения Пуассона
N_events = 1000
E_test = 0.01
N_vals = np.zeros(N_events, dtype=int)

for i in range(N_events):
    for directions in source.generate_photons(E_test):
        N_vals[i] += len(directions)

print(f"Среднее число фотонов: {np.mean(N_vals):.2f}")
print(f"Ожидаемое число фотонов: {Y * E_test:.2f}")
print(f"Стандартное отклонение: {np.std(N_vals):.2f}")
print(f"Теоретическое отклонение: {np.sqrt(Y * E_test):.2f}")
```
